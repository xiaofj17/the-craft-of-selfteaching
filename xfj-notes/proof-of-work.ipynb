#Github，Git，JupyterLab的关系

说实话，第一次看到老师写的那些具体步骤时，我当时就产生了逃避心理，因为基本上每一行都不太懂。

第二次硬着头皮把每一行的操作都执行了之后，脑细胞几近于死光不说，还根本不了解其中的逻辑关系。不了解逻辑关系，就不知道如何使用，那对后面的学习几乎就充满了绝望。

这一次重新开始之后，我啃下来了。

Github的功能非常多，在这里相当于一个线上的仓库。笑来老师把the-craft-of-selfteaching这本书放在这个大仓库里，我们在github上创建一个自己的账号后，就把笑来老师的这个the-craft-of-selfteaching复制（fork）到我们自己的仓库里。

但此时这本书还是在线上，使用上还是有些不太方便，所以要想办法将线上这本书放到本地，并且还需要下载到本地的这本书还能随时被JupyterLab调用。

这个时候就用上了Git，Git的功能强大，我还没有系统学习过他的使用方法，但我知道它是一个命令行的工具，可以将github上的书籍完整的clone到本地使用 。并且，现在它有图形界面使用的替代工具Github Desktop。

而JupyterLab则是本地代码的允许环境，可以允许代码，也可以做笔记，并且内容可以随时同步到github desktop，这样方便记录自己的任何修改。我下载的时候不是直接下载的JupyterLab，而是Anaconda。可以把Anaconda理解为一个大的分工明确的工具库，而JupyterLab是其中的小教室，专门可以用来允许代码，代用本地的资源。

如此一来，我们就可以创建一个自己的github，只需要到笑来老师的github账号上找到the-craft-of-selfteaching，然后点击“fork”，就到自己的仓库里了；

然后下载一个“Github Desktop”，用自己的github账号登录，然后就直接点击“clone”就将我们自己github仓库里从笑来老师仓库里fork过来的the-craft-of-selfteaching“clone”到我们的电脑里了。

然后打开Anaconda，找到JupyterLab，点击“launch”，然后看根据左边菜单的提示，找到之前clone好的the-craft-of-selfteaching文件夹，打开就可以直接使用了。书里的代码可以直接用，有什么想法可以直接写下来。真的是非常方便。



| 工具                        | 作用                                 | 你可以怎么用它                        |
| ------------------------- | ---------------------------------- | ------------------------------ |
| **GitHub（网站）**            | 在线代码仓库，用于存放、分享、协作开发内容              | Fork 项目、在线查看、发布 Pull Request   |
| **GitHub Desktop**        | Git 的图形界面，帮助你同步本地与 GitHub 上的项目     | 把 Fork 的仓库下载到你电脑上（clone），再提交更改 |
| **Anaconda / JupyterLab** | 本地运行 Python 和 Jupyter Notebook 的环境 | 打开并运行 `.ipynb` 文件，边学边写笔记       |
